```python

import tkinter as tk
from tkinter import ttk, messagebox
import json
import os


root = tk.Tk()
root.title("My Telephone Book")
root.geometry("650x550")
root.configure(bg="#c2efcd")


FILE_NAME = "contacts.json"


tk.Label(
    root,
    text="Telephone Book",
    font=("Arial", 22, "bold"),
    bg="#c2efcd",
    fg="#240703"
).pack(pady=15)

form = tk.Frame(root, bg="#c2efcd")
form.pack(pady=5)

tk.Label(form, text="Full Name:", bg="#c2efcd",font=("arial")).grid(row=0, column=0, padx=5, pady=7, sticky="w")
name_entry = tk.Entry(form, width=30)
name_entry.grid(row=0, column=1, padx=5, pady=7)

tk.Label(form, text="Phone Number:", bg="#c2efcd",font=("arial")).grid(row=1, column=0, padx=5, pady=7, sticky="w")
phone_entry = tk.Entry(form, width=30)
phone_entry.grid(row=1, column=1, padx=5, pady=7)

tk.Label(form, text="Gmail:", bg="#c2efcd",font=("arial")).grid(row=2, column=0, padx=5, pady=7, sticky="w")
gmail_entry = tk.Entry(form, width=30)
gmail_entry.grid(row=2, column=1, padx=5, pady=7)


table = ttk.Treeview(
    root,
    columns=("name", "phone", "gmail"),
    show="headings"
)

table.heading("name", text="Full Name")
table.heading("phone", text="Phone Number")
table.heading("gmail", text="Gmail")

table.column("name", width=170)
table.column("phone", width=150)
table.column("gmail", width=250)

table.pack(fill="both", expand=True, padx=15, pady=15)


# خواندن اطلاعات ذخیره‌شده
def load_contacts():
    if os.path.exists(FILE_NAME):
        try:
            with open(FILE_NAME, "r", encoding="utf-8") as file:
                contacts = json.load(file)

            for contact in contacts:
                table.insert("", "end", values=(
                    contact["name"],
                    contact["phone"],
                    contact["gmail"]
                ))
        except (json.JSONDecodeError, KeyError):
            messagebox.showwarning(
                "Warning",
                "The saved contacts file could not be read."
            )



def save_contacts():
    contacts = []

    for item in table.get_children():
        name, phone, gmail = table.item(item, "values")
        contacts.append({
            "name": name,
            "phone": phone,
            "gmail": gmail
        })

    with open(FILE_NAME, "w", encoding="utf-8") as file:
        json.dump(contacts, file, ensure_ascii=False, indent=4)


def add_contact():
    name = name_entry.get().strip()
    phone = phone_entry.get().strip()
    gmail = gmail_entry.get().strip()

    if not name or not phone or not gmail:
        messagebox.showwarning(
            "Incomplete Information",
            "Please fill in all fields."
        )
        return

    if "@" not in gmail or "." not in gmail.split("@")[-1]:
        messagebox.showwarning(
            "Invalid Gmail",
            "Please enter a valid email address."
        )
        return

    table.insert("", "end", values=(name, phone, gmail))
    save_contacts()

    name_entry.delete(0, tk.END)
    phone_entry.delete(0, tk.END)
    gmail_entry.delete(0, tk.END)

    messagebox.showinfo("Success", "Contact saved successfully!")



tk.Button(
    root,
    text="Save Contact",
    command=add_contact,
    bg="#3b8055",
    fg="white",
    font=("Arial", 12, "bold")
).pack(pady=10)

# نمایش مخاطبان قبلی هنگام اجرای برنامه
load_contacts()

root.mainloop()
```
